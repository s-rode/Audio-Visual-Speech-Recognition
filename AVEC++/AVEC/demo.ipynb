# Load and test model on samples
Notebook may require a restart to accommodate for package installation

```python
import os
os.chdir('/content/AVEC/')

# Imports
import nnet
import torch
import torchvision
import torchaudio
import matplotlib.pyplot as plt
import IPython.display as ipd
import jiwer
```

```python
# Model Device
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print("Device: {}".format(device))
print()

# Create Model VO
checkpoint_vo = "callbacks/LRS23/VO/EffConfInterCTC/checkpoints_swa-equal-90-100.ckpt"
model_vo = nnet.VisualEfficientConformerInterCTC().to(device)
model_vo.compile()
model_vo.summary()
model_vo.load(checkpoint_vo)
model_vo.eval()
print()

# Create Model AO
checkpoint_ao = "callbacks/LRS23/AO/EffConfCTC/checkpoints_swa-equal-200-210.ckpt"
model_ao = nnet.AudioEfficientConformerInterCTC(interctc_blocks=[]).to(device)
model_ao.compile()
model_ao.summary()
model_ao.load(checkpoint_ao)
model_ao.eval()
print()

# Create Model AV
checkpoint_av = "callbacks/LRS23/AV/EffConfInterCTC/checkpoints_swa-equal-60-70.ckpt"
model_av = nnet.AudioVisualEfficientConformerInterCTC().to(device)
model_av.compile()
model_av.summary()
model_av.load(checkpoint_av)
model_av.eval()
print()
```

```python
# Load Sample
sample_path = "media/macron.mp4"
video, audio, infos = torchvision.io.read_video(sample_path)
print(video.shape, audio.shape, infos)
ipd.Video(sample_path, embed=True)
```

```python
# Detect and Crops Lips
lip_crop = nnet.LipDetectCrop(mean_face_landmarks_path="media/20words_mean_face.npy")
sample_path_crop = sample_path.replace(".mp4", "_crop.mp4")
if os.path.isfile(sample_path_crop):
  video_crop, _, _ = torchvision.io.read_video(sample_path_crop)
else:
  video_crop = lip_crop(video, verbose=1)
  torchvision.io.write_video(filename=sample_path_crop, video_array=video_crop, fps=infos["video_fps"], video_codec="libx264")
ipd.Video(sample_path_crop, embed=True, height=400, width=400)
```

```python
video_preprocessing = torchvision.transforms.Compose([
    torchvision.transforms.ConvertImageDtype(dtype=torch.float32),
    nnet.Permute(dims=(0, 3, 1, 2)),
    torchvision.transforms.Grayscale(),
    nnet.Permute(dims=(1, 0, 2, 3)),
    nnet.NormalizeVideo(mean=(0.5,), std=(0.5,)),
    torchvision.transforms.CenterCrop((88, 88)),
    nnet.Permute(dims=(1, 2, 3, 0))
])

audio_preprocessing = torchaudio.transforms.Resample(orig_freq=infos["audio_fps"], new_freq=16000)

# Preprocess
video_p = video_preprocessing(video_crop)
audio_p = audio_preprocessing(audio)[0]
video_p = nnet.align_video_to_audio(video_p, audio_p)
print(video_p.shape, audio_p.shape)
```

```python
# Create Inputs
inputs_vo = (video_p.unsqueeze(dim=0).to(device), torch.tensor([video_p.shape[0]]).to(device))

inputs_ao = (audio_p.unsqueeze(dim=0).to(device), torch.tensor([audio_p.shape[0]]).to(device))

inputs_av = (video_p.unsqueeze(dim=0).to(device), torch.tensor([video_p.shape[0]]).to(device), audio_p.unsqueeze(dim=0).to(device), torch.tensor([audio_p.shape[0]]).to(device))
```

```python
# Forward
outputs_vo = model_vo(inputs_vo)["outputs"]
outputs_ao = model_ao(inputs_ao)["outputs"]
outputs_av = model_av(inputs_av)["outputs"]
```

```python
# Decoding Params
beamsearch = True
tokenizer_path="media/tokenizerbpe256.model"
ngram_path="media/6gram_lrs23.arpa"
neural_config_path = "configs/LRS23/LM/GPT-Small-demo.py"
neural_checkpoint = "checkpoints_epoch_10_step_2860.ckpt"

# Decoder
if beamsearch:
    decoder = nnet.CTCBeamSearchDecoder(tokenizer_path=tokenizer_path, ngram_path=ngram_path, neural_config_path=neural_config_path, neural_checkpoint=neural_checkpoint)
else:
    decoder = nnet.CTCGreedySearchDecoder(tokenizer_path=tokenizer_path)
```

```python
# Decode
preds_vo = decoder(outputs_vo)[0]
preds_ao = decoder(outputs_ao)[0]
preds_av = decoder(outputs_av)[0]
```

```python
groundtruth = "by polluting the oceans not mitigating co2 emissions and destroying our biodiversity we are killing our planet let us face it there is no planet b"

# Show Predictions
print("Groundtruth:\n{}".format(groundtruth))
print()
print("Visual-only prediction: wer={:.2f}%\n{}".format(100 * jiwer.wer(groundtruth, preds_vo), preds_vo))
print()
print("Audio-only prediction: wer={:.2f}%\n{}".format(100 * jiwer.wer(groundtruth, preds_ao), preds_ao))
print()
print("Audio-visual prediction: wer={:.2f}%\n{}".format(100 * jiwer.wer(groundtruth, preds_av), preds_av))
print()
```
